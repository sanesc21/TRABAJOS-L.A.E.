```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression
from scipy import stats
```

```python
df = pd.read_excel("stores_lae_exam1.xlsx")

# Limpieza 
def clean_to_float(val):
    try:
        if pd.isna(val): return np.nan
        # Elimina todo lo que no sea número o punto
        s = "".join(c for c in str(val).replace(',', '.') if c.isdigit() or c == '.')
        # Maneja casos donde hay dos puntos por error (ej. 4..57)
        if s.count('.') > 1:
            parts = s.split('.')
            s = parts[0] + '.' + "".join(parts[1:])
        return float(s)
    except:
        return np.nan

# Limpiar columnas numéricas
cols_num = ['Sales', 'Advertising', 'Store_Size', 'Customer_Satisfaction_Score', 
            'Inventory_Turnover_Rate', 'Employee_Turnover_Rate', 'Competitors', 
            'Years_Active', 'Manager_Experience']

for col in cols_num:
    df[col] = df[col].apply(clean_to_float)
```

```python
# 3. Preparar los datos (Eliminar nulos y crear dummies)
df_model = pd.get_dummies(df, drop_first=True).dropna()

# FORZAR TODO A FLOAT64 (Esto quita el error de 'inv')
X = df_model.drop('Sales', axis=1).astype(float)
y = df_model['Sales'].astype(float)

# 4. Regresión con Scikit-Learn
lm = LinearRegression()
lm.fit(X, y)
```

```python
# P- Values 
params = np.append(lm.intercept_, lm.coef_)
predictions = lm.predict(X)

# Matriz X con una columna de 1s para el intercepto
new_X = np.append(np.ones((len(X), 1)), X, axis=1)
```

```python
# Cálculo de la varianza de los coeficientes
# MSE = Suma de residuos al cuadrado / grados de libertad
MSE = (np.sum((y - predictions)**2)) / (len(new_X) - len(new_X[0]))

# Error estándar de los coeficientes
var_b = MSE * (np.linalg.inv(np.dot(new_X.T, new_X)).diagonal())
sd_b = np.sqrt(var_b)
```

```python
ts_b = params / sd_b
p_values = [2 * (1 - stats.t.cdf(np.abs(i), (len(new_X) - len(new_X[0])))) for i in ts_b]

# 6. Tabla final
resultados = pd.DataFrame({
    'Variable': ['Intercepto'] + list(X.columns),
    'Coeficiente': params,
    'P-Value': p_values
})

print(resultados.round(4))
```
