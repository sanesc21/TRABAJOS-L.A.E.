# BOOTSTRAPING

```python
import pandas as pd
df= pd.read_excel("Motor Trend Car Road Tests.xlsx")
```

```python
df.head()
```

Regresión

```python
import pandas as pd
import statsmodels.api as sm

X = sm.add_constant(df[['hp', 'qsec']])
y = df['mpg']

model_ols = sm.OLS(y, X).fit()
print(model_ols.summary())
print("\nIntervalos de confianza 95%:")
print(model_ols.conf_int().rename(columns={0: "IC inf", 1: "IC sup"}))
```

Bootstrap

```python
from sklearn.utils import resample

boot_betas = []

for i in range(1000):
    df_boot = resample(df)
    X_b = sm.add_constant(df_boot[['hp', 'qsec']])
    y_b = df_boot['mpg']
    model_b = sm.OLS(y_b, X_b).fit()
    boot_betas.append(model_b.params)

boot_df = pd.DataFrame(boot_betas)

print("Media (x̄_β):")
print(boot_df.mean().round(4))
print("\nDesviación estándar (σ_β):")
print(boot_df.std().round(4))
print("\nIntervalos de confianza 95% (Bootstrap):")
print(boot_df.quantile([0.025, 0.975]))
```

Comparación

```python
comparacion = pd.DataFrame({
    "OLS β̂":       model_ols.params,
    "OLS IC inf":  model_ols.conf_int()[0],
    "OLS IC sup":  model_ols.conf_int()[1],
    "Boot β̄":      boot_df.mean(),
    "Boot IC inf": boot_df.quantile(0.025),
    "Boot IC sup": boot_df.quantile(0.975),
}).round(4)

print(comparacion)
```

Conclusión

La comparación muestra que, aunque los coeficientes son casi idénticos, el bootstrap es más confiable porque no asume que los datos son perfectos. Mientras que el método OLS sugiere que el tiempo de aceleración (qsec) no es importante, el bootstrap demuestra que sí afecta negativamente al consumo. De manera resumida el remuestreo ofrece una visión más realista y segura, confirmando que tanto la potencia como el peso reducen significativamente la eficiencia del combustible.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score
```

```python
df = pd.read_excel('Motor Trend Car Road Tests.xlsx')
```

```python
df.head()
```

```python
# Definir target y features
target = 'mpg'
features = [col for col in df.columns if col not in [target, 'model']]

# Configuración
n_iterations = 1000
n_features_to_select = 3
results_lr = []
results_rf = []

# Split 50/50
train_df, test_df = train_test_split(df, test_size=0.5, random_state=42)
```

```python
# --- RANDOM FOREST "MANUAL" (bagging de árboles) ---
def random_forest_manual(X_train, y_train, X_test, n_trees=50):
    preds = []
    n = len(X_train)
    
    for i in range(n_trees):
        # Bootstrap sample
        indices = np.random.choice(range(n), size=n, replace=True)
        X_sample = X_train.iloc[indices]
        y_sample = y_train.iloc[indices]
        
        # Árbol de decisión
        tree = DecisionTreeRegressor(max_depth=5, random_state=i)
        tree.fit(X_sample, y_sample)
        
        # Predicción
        preds.append(tree.predict(X_test))
    
    # Promedio final
    return np.mean(preds, axis=0)
```

```python
for i in range(n_iterations):

    

    selected_vars = list(np.random.choice(features, n_features_to_select, replace=False))

    

    X_train = train_df[selected_vars]

    X_test = test_df[selected_vars]

    y_train = train_df[target]

    y_test = test_df[target]

    

    # Regresión Lineal

    lr = LinearRegression()

    lr.fit(X_train, y_train)

    y_pred_lr = lr.predict(X_test)

    results_lr.append(r2_score(y_test, y_pred_lr))

    

    # Bagging tipo Random Forest

    y_pred_rf = random_forest_manual(X_train, y_train, X_test, n_trees=50)

    results_rf.append(r2_score(y_test, y_pred_rf))

# Resultados

res_df = pd.DataFrame({

    'Regresión Lineal': results_lr,

    'Bagging (tipo RF)': results_rf

})

print(res_df.describe())
```

# RANDOM FOREST

```python
import pandas as pd
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import GridSearchCV, KFold


df = pd.read_excel("Motor Trend Car Road Tests.xlsx")
X = df.drop(columns=['model', 'mpg'])
y = df['mpg']
```

```python

kf = KFold(n_splits=10, shuffle=True, random_state=42)

# Definición de la "rejilla" (Grid) de parámetros
# El modelo probará TODAS las combinaciones de estos valores.
param_grid = {
    'n_estimators': [50, 100, 150, 200],
    'max_depth': [3, 5, 10, None],
    'max_features': [0.7, 0.8, 1.0],
    'min_samples_split': [2, 5]
}

# Configuración de GridSearchCV
rf = RandomForestRegressor(random_state=42)

grid_search = GridSearchCV(
    estimator=rf,
    param_grid=param_grid,
    cv=kf,
    scoring='r2',
    n_jobs=-1,  # Usa todos tus procesadores para ir más rápido
    verbose=1
)

# Entrenamiento
print("Iniciando búsqueda de la mejor combinación...")
grid_search.fit(X, y)

# Resultados 
print("-" * 30)
print(f"Mejor R2 detectado: {grid_search.best_score_:.4f}")
print(f"Mejores hiperparámetros:")
for param, value in grid_search.best_params_.items():
    print(f"  {param}: {value}")
print("-" * 30)
```

# BOOSTING

```python
import pandas as pd
from sklearn.ensemble import GradientBoostingRegressor
from sklearn.model_selection import GridSearchCV, KFold

df = pd.read_excel("Motor Trend Car Road Tests.xlsx")
X = df.drop(columns=['model', 'mpg'])
y = df['mpg']
```

```python
kf = KFold(n_splits=10, shuffle=True, random_state=42)

# Definición de la "rejilla" para Gradient Boosting
# Añadimos 'learning_rate' y ajustamos 'max_depth' (que suele ser menor en boosting)
param_grid = {
    'n_estimators': [100, 200, 300],
    'learning_rate': [0.01, 0.05, 0.1],
    'max_depth': [3, 4, 5],
    'subsample': [0.8, 1.0],
    'max_features': ['sqrt', 'log2', None]
}

# Configuración de GridSearchCV con el nuevo modelo
gbr = GradientBoostingRegressor(random_state=42)

grid_search = GridSearchCV(
    estimator=gbr,
    param_grid=param_grid,
    cv=kf,
    scoring='r2',
    n_jobs=-1,
    verbose=1
)

# Entrenamiento
print("Iniciando búsqueda de la mejor combinación para Gradient Boosting...")
grid_search.fit(X, y)

# Resultados 
print("-" * 30)
print(f"Mejor R2 detectado: {grid_search.best_score_:.4f}")
print("Mejores hiperparámetros:")
for param, value in grid_search.best_params_.items():
    print(f"  {param}: {value}")
print("-" * 30)
```
