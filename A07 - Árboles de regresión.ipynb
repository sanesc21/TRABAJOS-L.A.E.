# A07 - Árboles de regresión

En esta etapa se cargan la librerías y leemos el archivo de publicidad para definir nuestras variables. Usaremos la inversión en TV como predictor para las Ventas (Sales).

```python
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score

df = pd.read_csv('Advertising.csv')
df.head()
```

```python
X = df[["TV", "radio", "newspaper"]]
y = df['sales']
```

Comparamos qué tan bien se adaptan una regresión lineal y un árbol de decisión usando el 100% de la información. Esto nos sirve para ver la capacidad de ajuste inicial de cada modelo.

```python
#Regresión Lineal Simple
modelo_lr = LinearRegression()
modelo_lr.fit(X, y)
r2_lineal_total = r2_score(y, modelo_lr.predict(X))

# Arbol de Decisión
modelo_tree = DecisionTreeRegressor(random_state=42)
modelo_tree.fit(X, y)
r2_tree_total = r2_score(y, modelo_tree.predict(X))

print(f"R2 Lineal (Total): {r2_lineal_total:.4f}")
print(f"R2 Árbol (Total):  {r2_tree_total:.4f}")
```

Para probar si los modelos realmente aprenden o solo memorizan, dividimos los datos a la mitad. Entrenamos con una parte y evaluamos con la otra para medir su precisión real con datos desconocidos.

```python
# 50/50 Train-Split test
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.5, random_state=42)
```

```python
# Regresión Lineal con el 50/50
modelo_lr_split = LinearRegression().fit(X_train, y_train)
y_pred_lr = modelo_lr_split.predict(X_test)
r2_lineal_split = r2_score(y_test, y_pred_lr)
r2_li

# Árbol con el 50/50
modelo_tree_split = DecisionTreeRegressor(random_state=42).fit(X_train, y_train)
y_pred_tree = modelo_tree_split.predict(X_test)
r2_tree_split = r2_score(y_test, y_pred_tree)


print(f"R2 Lineal (Split 50/50): {r2_lineal_split:.4f}")
print(f"R2 Árbol (Split 50/50):  {r2_tree_split:.4f}")
```

```python
lr = LinearRegression()

lr.fit(X_train, y_train)

r2_train_tree = lr.score(X_train, y_train)
r2_test_tree = lr.score(X_test, y_test)

print("R2 en train para lineal:", r2_train_tree)
print("R2 en test para lineal:", r2_test_tree)
```

```python
tree = DecisionTreeRegressor(random_state=42)


tree.fit(X_train, y_train)
r2_train_tree = tree.score(X_train, y_train)
r2_test_tree = tree.score(X_test, y_test)

print("R2 en train para tree:", r2_train_tree)
print("R2 en test para tree:", r2_test_tree)
```

El árbol de decisión sacó un 1.0 perfecto en entrenamiento porque memorizó los datos (sobreajuste). Aunque su resultado en la prueba es bueno (0.93), la regresión lineal es más equilibrada y realista. Por lo que el modelo lineal es preferible porque es más estable y no se engaña con datos aprendidos de memoria.
