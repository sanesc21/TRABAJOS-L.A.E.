1.- Dos equipos juegan una serie de juegos (4 de 7, el que gane 4 juegos gana). Cada equipo tiene un 50% de probabilidad de ganar cada juego (no empates). ¿Cuál es la probabilidad de que la serie se vaya a 7 juegos? (5/16)

```python
import random

def simular_serie(intentos=1000000):
    siete_juegos = 0

    for _ in range(intentos):
        a = b = juegos = 0

        while a < 4 and b < 4:
            if random.random() < 0.5:
                a += 1
            else:
                b += 1
            juegos += 1

        if juegos == 7:
            siete_juegos += 1

    return siete_juegos / intentos


p_sim = simular_serie()
p_teo = 5/16

print(f"Simulada: {p_sim:.4f}")
print(f"Teórica:  {p_teo:.4f}")
```

3.- Tres amigos diferentes te dijeron que está lloviendo, y cada persona tiene 1/3 de probabilidad de mentir. Si el pronóstico del clima dijo que la probabilidad de lluvia era 25%, ¿cuál es la probabilidad de que sí esté lloviendo? (8/11)

```python
import random

def simulacion_lluvia(intentos=1000000):
    dicen_llueve = 0
    llovia = 0

    for _ in range(intentos):
        real = random.random() < 0.25
        todos_dicen_si = True

        for _ in range(3):
            verdad = random.random() > 1/3
            dice = real if verdad else not real
            if not dice:
                todos_dicen_si = False

        if todos_dicen_si:
            dicen_llueve += 1
            if real:
                llovia += 1

    return llovia / dicen_llueve if dicen_llueve else 0


p = simulacion_lluvia()
print(f"Simulada: {p:.4f}")
print(f"Teórica:  {8/11:.4f}")
```

5.- Tienes diez dados justos. Si arrojas todos al mismo tiempo, ¿cuál es la probabilidad de que la suma de los dados sea múltiplo de 6?

```python
import random

def simulacion(intentos=1000000):
    exitos = 0

    for _ in range(intentos):
        suma = sum(random.randint(1,6) for _ in range(10))
        if suma % 6 == 0:
            exitos += 1

    return exitos / intentos


p = simulacion()
print(f"Simulada: {p:.4f}")
print(f"Teórica:  {1/6:.4f}")
```
